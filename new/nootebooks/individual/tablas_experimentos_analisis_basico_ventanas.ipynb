# Tablas comparativas de experimentos individuales

Este notebook contiene únicamente las tres tablas finales de los experimentos A, B y C usando las estadísticas obtenidas con etiquetas individuales P33/P66 por trabajador.

Las métricas son medias de los folds LOSO de los notebooks individuales de Logistic Regression, Random Forest, SVM y XGBoost.

```python
import numpy as np
import pandas as pd

pd.set_option('display.float_format', lambda value: f'{value:,.4f}')
```

```python
# Tabla 1: Experimento A - ECG + HR/RR + temporales
summary_A = pd.DataFrame([
    ['Logistic Regression', 16, 0.5773, 0.5753, 0.5288],
    ['Random Forest',       16, 0.5901, 0.5860, 0.5345],
    ['SVM',                 16, 0.5568, 0.5547, 0.5197],
    ['XGBoost',             16, 0.5792, 0.5752, 0.5211],
], columns=[
    'Modelo',
    'Feature_count',
    'Accuracy_mean',
    'Balanced_Accuracy_mean',
    'Macro_F1_mean',
]).sort_values('Accuracy_mean', ascending=False).reset_index(drop=True)
display(summary_A.round(4))
```

```python
# Tabla 2: Experimento B - HR/RR + temporales
summary_B = pd.DataFrame([
    ['Logistic Regression', 10, 0.4965, 0.4976, 0.4281],
    ['Random Forest',       10, 0.4810, 0.4789, 0.4132],
    ['SVM',                 10, 0.4606, 0.4643, 0.3905],
    ['XGBoost',             10, 0.4793, 0.4784, 0.4158],
], columns=[
    'Modelo',
    'Feature_count',
    'Accuracy_mean',
    'Balanced_Accuracy_mean',
    'Macro_F1_mean',
]).sort_values('Accuracy_mean', ascending=False).reset_index(drop=True)
display(summary_B.round(4))
```

```python
# Tabla 3: Experimento C - ECG
summary_C = pd.DataFrame([
    ['Logistic Regression', 6, 0.5248, 0.5239, 0.4881],
    ['Random Forest',       6, 0.5152, 0.5109, 0.4618],
    ['SVM',                 6, 0.5207, 0.5196, 0.4809],
    ['XGBoost',             6, 0.5316, 0.5290, 0.4956],
], columns=[
    'Modelo',
    'Feature_count',
    'Accuracy_mean',
    'Balanced_Accuracy_mean',
    'Macro_F1_mean',
]).sort_values('Accuracy_mean', ascending=False).reset_index(drop=True)
display(summary_C.round(4))
```
